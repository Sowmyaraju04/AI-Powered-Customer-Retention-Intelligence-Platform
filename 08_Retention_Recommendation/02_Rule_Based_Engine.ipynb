```python
# ==========================================================
# AI-POWERED CUSTOMER RETENTION INTELLIGENCE PLATFORM
#
# PHASE 8
# RULE-BASED RETENTION RECOMMENDATION ENGINE
#
# NOTEBOOK:
# 02_Rule_Based_Engine.ipynb
# ==========================================================
```

```python
# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

import os
import pandas as pd
import numpy as np

print("=" * 60)
print("LIBRARIES IMPORTED SUCCESSFULLY")
print("=" * 60)
```

```python
# ==========================================================
# PROJECT PATH
# ==========================================================

PROJECT_ROOT = os.path.abspath("..")

print(PROJECT_ROOT)
```

```python
# ==========================================================
# LOAD RECOMMENDATION DATASET
# ==========================================================

dataset_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed",
    "customer_recommendation_data.csv"
)

recommendation_data = pd.read_csv(dataset_path)

print("=" * 60)
print("RECOMMENDATION DATASET")
print("=" * 60)

print(recommendation_data.head())

print("\nShape")
print(recommendation_data.shape)
```

```python
# ==========================================================
# CUSTOMER VALUE SEGMENT
# ==========================================================

recommendation_data["Customer_Value"] = pd.qcut(
    recommendation_data["Monetary"],
    q=3,
    labels=[
        "Low Value",
        "Medium Value",
        "High Value"
    ]
)

print("=" * 60)
print("CUSTOMER VALUE SEGMENTS")
print("=" * 60)

print(
    recommendation_data["Customer_Value"]
    .value_counts()
)
```

```python
# ==========================================================
# RETENTION RECOMMENDATION ENGINE
# ==========================================================

def generate_recommendation(row):

    risk = row["Risk_Level"]
    review = row["average_review_score"]
    value = row["Customer_Value"]
    frequency = row["Frequency"]

    # Rule 1
    if risk == "High" and review <= 2 and value == "High Value":
        return "VIP Service Recovery"

    # Rule 2
    elif risk == "High" and review <= 2:
        return "Customer Support Follow-Up"

    # Rule 3
    elif risk == "High" and value == "High Value":
        return "Exclusive Loyalty Offer"

    # Rule 4
    elif risk == "High":
        return "Personalized Discount"

    # Rule 5
    elif risk == "Medium" and frequency <= 1:
        return "Re-engagement Email"

    # Rule 6
    elif risk == "Medium" and value == "High Value":
        return "Loyalty Reward"

    # Rule 7
    elif risk == "Medium":
        return "Personalized Promotion"

    # Rule 8
    elif risk == "Low" and value == "High Value":
        return "VIP Appreciation"

    # Rule 9
    elif risk == "Low":
        return "Cross-Sell Recommendation"

    return "No Action"


recommendation_data["Recommendation"] = (
    recommendation_data.apply(
        generate_recommendation,
        axis=1
    )
)

print("=" * 60)
print("RECOMMENDATIONS GENERATED")
print("=" * 60)

print(
    recommendation_data["Recommendation"]
    .value_counts()
)
```

```python
# ==========================================================
# CAMPAIGN TYPE
# ==========================================================

campaign_map = {

    "VIP Service Recovery":"Service",

    "Customer Support Follow-Up":"Service",

    "Exclusive Loyalty Offer":"Retention",

    "Personalized Discount":"Promotion",

    "Re-engagement Email":"Reactivation",

    "Loyalty Reward":"Loyalty",

    "Personalized Promotion":"Promotion",

    "VIP Appreciation":"Loyalty",

    "Cross-Sell Recommendation":"Growth",

    "No Action":"None"

}

recommendation_data["Campaign_Type"] = (
    recommendation_data["Recommendation"]
    .map(campaign_map)
)
```

```python
# ==========================================================
# BUSINESS REASON
# ==========================================================

def business_reason(row):

    return (
        f"Risk Level: {row['Risk_Level']}, "
        f"Review Score: {row['average_review_score']}, "
        f"Customer Value: {row['Customer_Value']}"
    )

recommendation_data["Business_Reason"] = (
    recommendation_data.apply(
        business_reason,
        axis=1
    )
)
```

```python
# ==========================================================
# EXPECTED OUTCOME
# ==========================================================

outcome_map = {

    "VIP Service Recovery":"Recover high-value customer",

    "Customer Support Follow-Up":"Improve customer satisfaction",

    "Exclusive Loyalty Offer":"Increase retention",

    "Personalized Discount":"Reduce churn risk",

    "Re-engagement Email":"Increase activity",

    "Loyalty Reward":"Increase loyalty",

    "Personalized Promotion":"Increase purchases",

    "VIP Appreciation":"Strengthen relationship",

    "Cross-Sell Recommendation":"Increase revenue",

    "No Action":"Monitor"

}

recommendation_data["Expected_Outcome"] = (
    recommendation_data["Recommendation"]
    .map(outcome_map)
)
```

```python
# ==========================================================
# FINAL DATASET
# ==========================================================

print("=" * 60)
print("FINAL RETENTION RECOMMENDATION DATASET")
print("=" * 60)

print(recommendation_data.head())

print("\nColumns")

print(recommendation_data.columns.tolist())
```

```python
# ==========================================================
# SAVE DATASET
# ==========================================================

output_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed",
    "customer_retention_recommendations.csv"
)

recommendation_data.to_csv(
    output_path,
    index=False
)

print("=" * 60)
print("RETENTION RECOMMENDATION DATASET SAVED")
print("=" * 60)

print(output_path)
```
