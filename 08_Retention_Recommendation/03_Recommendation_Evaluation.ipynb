```python
# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

import os
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 60)
print("LIBRARIES IMPORTED SUCCESSFULLY")
print("=" * 60)
```

```python
# ==========================================================
# PROJECT PATH
# ==========================================================

PROJECT_ROOT = os.path.abspath("..")

print(PROJECT_ROOT)
```

```python
# ==========================================================
# LOAD RETENTION RECOMMENDATIONS
# ==========================================================

dataset_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed",
    "customer_retention_recommendations.csv"
)

recommendation_data = pd.read_csv(dataset_path)

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print(recommendation_data.head())

print("\nShape")

print(recommendation_data.shape)
```

```python
# ==========================================================
# RECOMMENDATION DISTRIBUTION
# ==========================================================

recommendation_counts = (
    recommendation_data["Recommendation"]
    .value_counts()
)

print("=" * 60)
print("RECOMMENDATION DISTRIBUTION")
print("=" * 60)

print(recommendation_counts)

plt.figure(figsize=(10,6))

recommendation_counts.plot(kind="bar")

plt.title("Recommendation Distribution")

plt.xlabel("Recommendation")

plt.ylabel("Customers")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# CAMPAIGN TYPE DISTRIBUTION
# ==========================================================

campaign_counts = (
    recommendation_data["Campaign_Type"]
    .value_counts()
)

print("=" * 60)
print("CAMPAIGN DISTRIBUTION")
print("=" * 60)

print(campaign_counts)

plt.figure(figsize=(8,5))

campaign_counts.plot(kind="bar")

plt.title("Campaign Type Distribution")

plt.xlabel("Campaign")

plt.ylabel("Customers")

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# PRIORITY DISTRIBUTION
# ==========================================================

priority_counts = (
    recommendation_data["Priority"]
    .value_counts()
)

print("=" * 60)
print("PRIORITY DISTRIBUTION")
print("=" * 60)

print(priority_counts)

plt.figure(figsize=(6,5))

priority_counts.plot(kind="bar")

plt.title("Business Priority Distribution")

plt.xlabel("Priority")

plt.ylabel("Customers")

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# CUSTOMER VALUE VS RECOMMENDATION
# ==========================================================

value_recommendation = pd.crosstab(
    recommendation_data["Customer_Value"],
    recommendation_data["Recommendation"]
)

print("=" * 60)
print("CUSTOMER VALUE VS RECOMMENDATION")
print("=" * 60)

print(value_recommendation)

value_recommendation.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Customer Value vs Recommendation")

plt.ylabel("Customers")

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# RISK LEVEL VS RECOMMENDATION
# ==========================================================

risk_recommendation = pd.crosstab(
    recommendation_data["Risk_Level"],
    recommendation_data["Recommendation"]
)

print("=" * 60)
print("RISK LEVEL VS RECOMMENDATION")
print("=" * 60)

print(risk_recommendation)

risk_recommendation.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Risk Level vs Recommendation")

plt.ylabel("Customers")

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# RECOMMENDATION SUMMARY
# ==========================================================

summary = pd.DataFrame({

    "Metric":[

        "Total Customers",

        "Unique Recommendations",

        "Campaign Types",

        "High Priority Customers"

    ],

    "Value":[

        len(recommendation_data),

        recommendation_data["Recommendation"].nunique(),

        recommendation_data["Campaign_Type"].nunique(),

        (recommendation_data["Priority"]=="Critical").sum()

    ]

})

print("=" * 60)
print("EXECUTIVE SUMMARY")
print("=" * 60)

print(summary)
```
