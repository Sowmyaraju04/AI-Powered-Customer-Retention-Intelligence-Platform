## AI-Powered Customer Retention Intelligence Platform

### Objective

The objective of this phase is to develop a machine learning model capable of identifying customers at risk of inactivity based on their historical purchasing behavior, customer value, satisfaction, and engagement characteristics.

The modeling target is an inactivity-based churn proxy created during the feature engineering phase.

```python
# ==========================================================
# LOAD MODELING DATASET
# ==========================================================

model_path = "data/processed/customer_retention_model_data.csv"

model_data = pd.read_csv(model_path)

print("=" * 60)
print("MODEL DATASET LOADED")
print("=" * 60)

print("\nShape:")
print(model_data.shape)

print("\nColumns:")
print(model_data.columns.tolist())
```

```python
# ==========================================================
# DATASET INSPECTION
# ==========================================================

print("=" * 60)
print("DATASET INSPECTION")
print("=" * 60)

print("\nData Types:")
print(model_data.dtypes)

print("\nMissing Values:")
print(model_data.isnull().sum())

print("\nDuplicate Rows:")
print(model_data.duplicated().sum())
```

```python
# ==========================================================
# TARGET DISTRIBUTION
# ==========================================================

print("=" * 60)
print("TARGET DISTRIBUTION")
print("=" * 60)

print("\nCounts:")
print(
    model_data["churn_label"]
    .value_counts()
)

print("\nPercentage:")
print(
    model_data["churn_label"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ==========================================================
# DEFINE TARGET
# ==========================================================

target_column = "churn_label"

y = model_data[target_column]

print("Target Variable:")
print(target_column)

print("\nTarget Shape:")
print(y.shape)
```

```python
# ==========================================================
# DEFINE PREDICTIVE FEATURES
# ==========================================================

feature_columns = [
    "Frequency",
    "Monetary",
    "average_review_score",
    "preferred_payment_method",
    "average_order_value",
    "customer_lifetime_value"
]

X = model_data[feature_columns].copy()

print("=" * 60)
print("PREDICTIVE FEATURES")
print("=" * 60)

print("\nFeatures:")
for feature in feature_columns:
    print("-", feature)

print("\nFeature Shape:")
print(X.shape)
```

```python
feature_columns = [
    "Frequency",
    "Monetary",
    "average_review_score",
    "preferred_payment_method",
    "average_order_value"
]
```

```python
X = model_data[feature_columns].copy()
```

```python
# ==========================================================
# TRAIN / TEST SPLIT
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("=" * 60)
print("TRAIN / TEST SPLIT")
print("=" * 60)

print("\nTraining Features:", X_train.shape)
print("Testing Features :", X_test.shape)

print("\nTraining Target:")
print(y_train.value_counts(normalize=True))

print("\nTesting Target:")
print(y_test.value_counts(normalize=True))
```
