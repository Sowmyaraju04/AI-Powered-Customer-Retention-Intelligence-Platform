```python
# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

import os
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 60)
print("LIBRARIES IMPORTED SUCCESSFULLY")
print("=" * 60)
```

```python
# ==========================================================
# PROJECT ROOT
# ==========================================================

PROJECT_ROOT = os.path.abspath("..")

print(PROJECT_ROOT)
```

```python
# ==========================================================
# LOAD AI CUSTOMER EXPLANATIONS
# ==========================================================

dataset_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "final",
    "ai_customer_explanations.csv"
)

ai_data = pd.read_csv(dataset_path)

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print(ai_data.head())

print("\nShape")

print(ai_data.shape)
```

```python
# ==========================================================
# EXECUTIVE KPI SUMMARY
# ==========================================================

executive_summary = pd.DataFrame({

    "Metric":[

        "Total Customers",

        "High Risk Customers",

        "Average Risk Probability",

        "Average Review Score",

        "Average Customer Spend",

        "Recommendation Types",

        "Campaign Types"

    ],

    "Value":[

        len(ai_data),

        (ai_data["Risk_Level"]=="High").sum(),

        round(ai_data["Risk_Probability"].mean(),3),

        round(ai_data["average_review_score"].mean(),2),

        round(ai_data["Monetary"].mean(),2),

        ai_data["Recommendation"].nunique(),

        ai_data["Campaign_Type"].nunique()

    ]

})

print("="*60)
print("EXECUTIVE KPI SUMMARY")
print("="*60)

print(executive_summary)
```

```python
# ==========================================================
# AI EXECUTIVE INSIGHTS
# ==========================================================

executive_insights = [

    f"Total customers analysed: {len(ai_data):,}.",

    f"{(ai_data['Risk_Level']=='High').sum():,} customers are classified as High Risk.",

    f"Average customer spend is {ai_data['Monetary'].mean():.2f}.",

    f"Average customer review score is {ai_data['average_review_score'].mean():.2f}.",

    f"The most common campaign is '{ai_data['Campaign_Type'].mode()[0]}'.",

    f"The most common recommendation is '{ai_data['Recommendation'].mode()[0]}'."

]

print("="*60)
print("AI EXECUTIVE INSIGHTS")
print("="*60)

for i, insight in enumerate(executive_insights,1):

    print(f"{i}. {insight}")
```

```python
# ==========================================================
# RISK LEVEL DISTRIBUTION
# ==========================================================

plt.figure(figsize=(7,5))

ai_data["Risk_Level"].value_counts().plot(kind="bar")

plt.title("Customer Risk Level Distribution")

plt.xlabel("Risk Level")

plt.ylabel("Customers")

plt.tight_layout()

plt.show()
```

```python
# ==========================================================
# SAVE EXECUTIVE SUMMARY
# ==========================================================

summary_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "final",
    "executive_summary.csv"
)

executive_summary.to_csv(
    summary_path,
    index=False
)

print("="*60)
print("EXECUTIVE SUMMARY SAVED")
print("="*60)

print(summary_path)
```
