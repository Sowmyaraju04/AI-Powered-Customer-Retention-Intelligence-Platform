```python
# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

import os
import pandas as pd
import numpy as np

print("=" * 60)
print("LIBRARIES IMPORTED SUCCESSFULLY")
print("=" * 60)
```

```python
# ==========================================================
# PROJECT ROOT
# ==========================================================

PROJECT_ROOT = os.path.abspath("..")

print("=" * 60)
print("PROJECT ROOT")
print("=" * 60)

print(PROJECT_ROOT)
```

```python
# ==========================================================
# LOAD RETENTION RECOMMENDATION DATASET
# ==========================================================

dataset_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed",
    "customer_retention_recommendations.csv"
)

ai_data = pd.read_csv(dataset_path)

print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print(ai_data.head())

print("\nShape")

print(ai_data.shape)
```

```python
# ==========================================================
# AI FEATURE SELECTION
# ==========================================================

ai_dataset = ai_data[
    [
        "customer_unique_id",
        "Frequency",
        "Monetary",
        "average_order_value",
        "average_review_score",
        "preferred_payment_method",
        "Risk_Probability",
        "Risk_Level",
        "Priority",
        "Customer_Value",
        "Recommendation",
        "Campaign_Type",
        "Business_Reason",
        "Expected_Outcome"
    ]
].copy()

print("=" * 60)
print("AI DATASET CREATED")
print("=" * 60)

print(ai_dataset.head())

print("\nColumns")

print(ai_dataset.columns.tolist())
```

```python
# ==========================================================
# CREATE AI INSIGHT TAGS
# ==========================================================

def create_insight_tag(row):

    tags = []

    if row["Risk_Level"] == "High":
        tags.append("High Churn Risk")

    if row["average_review_score"] <= 2:
        tags.append("Low Satisfaction")

    if row["Customer_Value"] == "High Value":
        tags.append("High Value Customer")

    if row["Frequency"] <= 1:
        tags.append("Low Purchase Frequency")

    if row["Monetary"] >= 500:
        tags.append("High Spending")

    if len(tags) == 0:
        tags.append("Stable Customer")

    return ", ".join(tags)

ai_dataset["AI_Insight_Tags"] = ai_dataset.apply(
    create_insight_tag,
    axis=1
)

print("=" * 60)
print("AI INSIGHT TAGS CREATED")
print("=" * 60)

print(ai_dataset["AI_Insight_Tags"].head())
```

```python
# ==========================================================
# AI READINESS SUMMARY
# ==========================================================

summary = pd.DataFrame({

    "Metric":[
        "Total Customers",
        "Total Features",
        "Risk Levels",
        "Recommendation Types",
        "Campaign Types"
    ],

    "Value":[
        len(ai_dataset),
        ai_dataset.shape[1],
        ai_dataset["Risk_Level"].nunique(),
        ai_dataset["Recommendation"].nunique(),
        ai_dataset["Campaign_Type"].nunique()
    ]

})

print("=" * 60)
print("AI DATASET SUMMARY")
print("=" * 60)

print(summary)
```

```python
# ==========================================================
# SAVE AI DATASET
# ==========================================================

final_folder = os.path.join(
    PROJECT_ROOT,
    "data",
    "final"
)

os.makedirs(
    final_folder,
    exist_ok=True
)

output_path = os.path.join(
    final_folder,
    "ai_ready_dataset.csv"
)

ai_dataset.to_csv(
    output_path,
    index=False
)

print("=" * 60)
print("AI READY DATASET SAVED")
print("=" * 60)
print(output_path)
```
