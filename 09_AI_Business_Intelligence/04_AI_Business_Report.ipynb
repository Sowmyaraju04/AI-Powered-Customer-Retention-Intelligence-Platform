```python
# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

import os
import pandas as pd

print("=" * 60)
print("LIBRARIES IMPORTED SUCCESSFULLY")
print("=" * 60)
```

```python
# ==========================================================
# PROJECT ROOT
# ==========================================================

PROJECT_ROOT = os.path.abspath("..")

print(PROJECT_ROOT)
```

```python
# ==========================================================
# LOAD AI DATASETS
# ==========================================================

customer_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "final",
    "ai_customer_explanations.csv"
)

summary_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "final",
    "executive_summary.csv"
)

customer_data = pd.read_csv(customer_path)
executive_summary = pd.read_csv(summary_path)

print("=" * 60)
print("DATASETS LOADED")
print("=" * 60)

print(customer_data.shape)
print(executive_summary.shape)
```

```python
# ==========================================================
# GENERATE BUSINESS REPORT
# ==========================================================

business_report = {

    "Total_Customers": len(customer_data),

    "High_Risk_Customers":
    (customer_data["Risk_Level"]=="High").sum(),

    "Average_Risk_Probability":
    round(customer_data["Risk_Probability"].mean(),3),

    "Average_Review_Score":
    round(customer_data["average_review_score"].mean(),2),

    "Average_Customer_Spend":
    round(customer_data["Monetary"].mean(),2),

    "Most_Common_Recommendation":
    customer_data["Recommendation"].mode()[0],

    "Most_Common_Campaign":
    customer_data["Campaign_Type"].mode()[0]
}

report_df = pd.DataFrame(
    business_report.items(),
    columns=["Metric","Value"]
)

print("=" * 60)
print("BUSINESS REPORT")
print("=" * 60)

print(report_df)
```

```python
# ==========================================================
# EXECUTIVE BUSINESS NARRATIVE
# ==========================================================

narrative = f"""
Executive Business Report

The analysis covered {len(customer_data):,} customers.

High-risk customers identified:
{(customer_data['Risk_Level']=='High').sum():,}

Average customer spend:
{customer_data['Monetary'].mean():.2f}

Average review score:
{customer_data['average_review_score'].mean():.2f}

Primary campaign recommendation:
{customer_data['Campaign_Type'].mode()[0]}

Most common retention strategy:
{customer_data['Recommendation'].mode()[0]}

Business Recommendation:

Focus immediate retention efforts on high-risk customers while continuing loyalty initiatives for valuable low-risk customers. Monitor campaign performance and customer satisfaction to improve long-term retention.
"""

print("=" * 60)
print("AI BUSINESS REPORT")
print("=" * 60)

print(narrative)
```

```python
# ==========================================================
# SAVE BUSINESS REPORT
# ==========================================================

output_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "final",
    "business_report_data.csv"
)

report_df.to_csv(
    output_path,
    index=False
)

print("=" * 60)
print("BUSINESS REPORT SAVED")
print("=" * 60)

print(output_path)
```
